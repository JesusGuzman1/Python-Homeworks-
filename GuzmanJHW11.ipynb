5. Write a ``ChargedParticle`` class that inherits from the ``Particle`` class, but also has an attribute for the charge of the particle.


```python
import numpy as np
class particle(object):
    def __init__(self, pos, vel, m):
        self.x , self.y, self.z = pos 
        self.vx, self.vy, self.vz = vel
        self.m = m 

    def ke(self):
        return self.m*(self.vx**2+self.vy**2+self.vz**2)/2
    def distance(self,particle2):
        return np.sqrt((self.x-particle2.x)**2+(self.y-particle2.y)**2+(self.z-particle2.z)**2)
    def newp(self,time):
        return [self.x+self.vx*time,self.y+self.vy**time,self.z+self.vz**time]



# part 5
class ChargedParticle(particle):
    
    def __init__(self, pos, vel,m, charge):
    
        super().__init__(pos,vel,m)
        #charge
        self.charge = charge
        
p1 = particle([2,4,-3],[1,1,1],5)
p2 = particle([3,-2,1],[1,1,1],5)

print(p1.ke())
print(p1.distance(p2))
print(p1.newp(2))
chargedp = ChargedParticle([2,4,-3],[1,1,1],5,2)
chargedp2 = ChargedParticle([3,-2,1],[1,1,1],5,3)
print(chargedp.charge)
print(chargedp2.charge)
```
